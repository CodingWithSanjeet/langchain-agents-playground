# 🦜🔗 LangChain Study Notes: Multi-Provider LLM Integration

## 📌 Quick Reference & Overview
LangChain provides a unified interface to interact with various Large Language Model (LLM) providers like **OpenAI**, **Google Gemini**, and **Groq**.

### 🔑 Key Concepts Covered:
1. **Unified Factory Pattern (`init_chat_model`)**: One-line model initialization format (`"provider:model-name"` or `"model-name"`) allowing model swapping without changing import statements.
2. **Provider-Specific Classes**: Direct integration using dedicated packages:
   - `langchain_openai.ChatOpenAI`
   - `langchain_google_genai.ChatGoogleGenerativeAI`
   - `langchain_groq.ChatGroq`
3. **Standardized Invocation**: All chat models implement `.invoke()`, returning an `AIMessage` containing `.content`, metadata, and token usage statistics.

---

### 🛠️ Step 1: Environment Variables Setup
Load API keys from local `.env` file into `os.environ` for all three providers:
- `OPENAI_API_KEY`: Required for OpenAI GPT models.
- `GOOGLE_API_KEY`: Required for Google Gemini models.
- `GROQ_API_KEY`: Required for Groq cloud inference.

```python
import os
from dotenv import load_dotenv
load_dotenv()

os.environ["OPENAI_API_KEY"]=os.getenv("OPENAI_API_KEY")
os.environ["GOOGLE_API_KEY"]=os.getenv("GOOGLE_API_KEY")
os.environ["GROQ_API_KEY"]=os.getenv("GROQ_API_KEY")
```

---
### 🤖 Section 1: OpenAI Model Integration

#### Method A: Unified Factory Pattern (`init_chat_model`)
- `init_chat_model("gpt-4.1")` automatically detects OpenAI from model name.
- `.invoke(prompt)` returns an `AIMessage` object.
- Access the text response via `response.content`.

#### Method B: Direct Class (`ChatOpenAI`)
- Imported from `langchain_openai` for explicit control over OpenAI parameters.

```python
from langchain.chat_models import init_chat_model

model = init_chat_model("gpt-4.1")
model
```

```python
### invoke the model
response=model.invoke("Hello, How are you?")
response
```

```python
response.content
```

```python
### Chat OpenAI
from langchain_openai import ChatOpenAI
model = ChatOpenAI(model="gpt-4.1")
response=model.invoke("Hello, how are you?")
response
```

---
### 🌐 Section 2: Google Gemini Model Integration

#### Method A: `init_chat_model` with Provider Prefix
- Prefix format: `init_chat_model("google_genai:gemini-3.5-flash-lite")` explicitly targets Google GenAI.

#### Method B: Direct Class (`ChatGoogleGenerativeAI`)
- Imported from `langchain_google_genai` package (`ChatGoogleGenerativeAI(model="gemini-3.5-flash-lite")`).

```python
import os
from langchain.chat_models import init_chat_model
os.environ["GOOGLE_API_KEY"]= os.getenv("GOOGLE_API_KEY")
model = init_chat_model("google_genai:gemini-3.5-flash-lite")
response = model.invoke("Why do parrots talk?")
response.content
```

```python
from langchain_google_genai import ChatGoogleGenerativeAI

model = ChatGoogleGenerativeAI(model="gemini-3.5-flash-lite")
response = model.invoke("Hi, How are you?")
response
```

---
### ⚡ Section 3: Groq Model Integration (Ultra-Fast Open Weights)

#### Method A: `init_chat_model` with Provider Prefix
- Prefix format: `init_chat_model("groq:qwen/qwen3.8-27b")` routes inference to Groq's high-speed LPU infrastructure.

#### Method B: Direct Class (`ChatGroq`)
- Imported from `langchain_groq` (`ChatGroq(model="qwen/qwen3.8-27b")`).

```python
import os
from langchain.chat_models import init_chat_model
os.environ["GROQ_API_KEY"]= os.getenv("GROQ_API_KEY")
model = init_chat_model("groq:qwen/qwen3.8-27b")
response = model.invoke("Why do parrots talk?")
response.content
```

```python
from langchain_groq import ChatGroq

model = ChatGroq(model="qwen/qwen3.8-27b")
response = model.invoke("Hi, How are you?")
response
```
